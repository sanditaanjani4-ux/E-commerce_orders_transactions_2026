# Latihan SKLearn SVM untuk Klasifikasi

```python
import pandas as pd

df = pd.read_csv('diabetes.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
# memisahkan atribut pada dataset dan menyimpannya pada sebuah variabel
X = df[df.columns[:8]]

# memisahkan label pada dataset dan menyimpannya pada sebuah variabel
y = df['Outcome']
```

```python
from sklearn.preprocessing import StandardScaler

# standarisasi nilai-nilai dari dataset
scaler = StandardScaler()
scaler.fit(X)
X = scaler.transform(X)
```

```python
from sklearn.model_selection import train_test_split

# memisahkan data untuk training dan testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.33, random_state=42)
```

```python
from sklearn.svm import SVC

# membuat objek SVC dan memanggil fungsi fit untuk melatih model
clf = SVC()
clf.fit(X_train, y_train)
```

```python
# Menampilkan skor akurasi prediksi
clf.score(X_test, y_test)
```

```python
import pandas as pd

# membaca dataset dan mengubahnya menjadi dataframe
data = pd.read_csv('diabetes.csv')
```

```python
data.info()
```

```python
data.head()
```

```python
import numpy as np

# memisahkan atribut dan label
X = data.drop('Outcome', axis=1)
y = data['Outcome']

# mengubah bentuk atribut
X = np.array(X)
X = X[:, :]

print("Atribut berhasil dipisahkan")
print("Jumlah data:", X.shape)
print("Jumlah label:", y.shape)
```

```python
from sklearn.svm import SVC

# membangun model dengan parameter C, gamma, dan kernel
model = SVC(C=1000, gamma=0.05, kernel='rbf')

# melatih model dengan fungsi fit
model.fit(X, y)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Dataset points
x1 = np.array([2, 1, 3, 1, 4, 2, 5, 2])  # Study Hours
x2 = np.array([8, 6, 7, 4, 6, 5, 7, 3])  # Sleep Hours
y = np.array([1, -1, 1, -1, 1, -1, 1, -1]) # Passed Exam

# Equation of the decision boundary and margins:
# Hyperplane: x2 = -x1 + 8.5
# Positive Margin: x2 = -x1 + 9
# Negative Margin: x2 = -x1 + 8

x1_range = np.linspace(0, 6, 100) # Range of x1 values for plotting
decision_boundary = -x1_range + 8.5
margin_positive = -x1_range + 9
margin_negative = -x1_range + 8

# Plotting the dataset points
plt.figure(figsize=(8, 6))
plt.scatter(x1[y == 1], x2[y == 1], color='blue', label='Passed (y=1)', s=100)
plt.scatter(x1[y == -1], x2[y == -1], color='red', label='Failed (y=-1)', s=100)

# Plotting the hyperplane and margins
plt.plot(x1_range, decision_boundary, color='green', label='Decision Boundary', linewidth=2)
plt.plot(x1_range, margin_positive, color='orange', linestyle='--', label='Positive Margin', linewidth=1.5)
plt.plot(x1_range, margin_negative, color='purple', linestyle='--', label='Negative Margin', linewidth=1.5)

# Labels and grid
plt.title('SVM Decision Boundary and Margins', fontsize=14)
plt.xlabel('Study Hours (x1)', fontsize=12)
plt.ylabel('Sleep Hours (x2)', fontsize=12)
plt.axhline(0, color='black', linewidth=0.5, linestyle='--')
plt.axvline(0, color='black', linewidth=0.5, linestyle='--')
plt.grid(color='gray', linestyle='--', linewidth=0.5, alpha=0.7)
plt.legend(fontsize=12)
plt.ylim(0, 10)
plt.show()
```

# Library Sklearn

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import svm

# Dataset (same as the one we used earlier)
x1 = np.array([2, 1, 3, 1, 4, 2, 5, 2]) # Study Hours
x2 = np.array([8, 6, 7, 4, 6, 5, 7, 3]) # Sleep Hours
y = np.array([1, -1, 1, -1, 1, -1, 1, -1]) # Passed Exam

# Reshaping data to match the required input format
X = np.column_stack((x1, x2))

# Create and fit the SVM model
clf = svm.SVC(kernel='linear') # Linear kernel
clf.fit(X, y)

# Get the decision boundary (hyperplane)
w = clf.coef_[0]
b = clf.intercept_[0]

# Plotting the dataset points
plt.figure(figsize=(8, 6))
plt.scatter(x1[y == 1], x2[y == 1], color='blue', label='Passed (y=1)', s=100)
plt.scatter(x1[y == -1], x2[y == -1], color='red', label='Failed (y=-1)', s=100)

# Plot decision boundary
x1_range = np.linspace(0, 6, 100)
decision_boundary = -(w[0] * x1_range + b) / w[1]
plt.plot(x1_range, decision_boundary, color='green', label='Decision Boundary', linewidth=2)

plt.title('SVM Decision Boundary (Scikit-learn)', fontsize=14)
plt.xlabel('Study Hours (x1)', fontsize=12)
plt.ylabel('Sleep Hours (x2)', fontsize=12)
plt.axhline(0, color='black', linewidth=0.5, linestyle='--')
plt.axvline(0, color='black', linewidth=0.5, linestyle='--')
plt.grid(color='gray', linestyle='--', linewidth=0.5, alpha=0.7)
plt.legend(fontsize=12)
plt.ylim(0, 10)
plt.show()
```
