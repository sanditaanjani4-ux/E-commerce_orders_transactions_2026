# identifikasi dataset Fake And Real News

```python
df_fake = pd.read_csv("FakeAndRealNews/Fake.csv")
df_true = pd.read_csv("FakeAndRealNews/True.csv")

print("DATA FAKE")
print("Jumlah data dan variabel:", df_fake.shape)
print("Nama kolom:", df_fake.columns.tolist())
display(df_fake.head())
df_fake.info()

print("\nDATA TRUE")
print("Jumlah data dan variabel:", df_true.shape)
print("Nama kolom:", df_true.columns.tolist())
display(df_true.head())
df_true.info()
```

```python
df_news = pd.concat([df_fake, df_true], ignore_index=True)

df_news["label"] = ["Fake"] * len(df_fake) + ["Real"] * len(df_true)

print("JUMLAH DATA DAN VARIABEL")
print(df_news.shape)

print("\nDATA FITUR:")
print(df_news.drop(columns=["label"]).columns.tolist())

print("\nDATA TARGET:")
print("label")

print("\nJUMLAH LABEL:")
print(df_news["label"].value_counts())

display(df_news.head())
```

```python
import json
import pandas as pd

data = []

with open("ImageTags/face_detection.json", "r", encoding="utf-8") as file:
    for baris in file:
        if baris.strip():
            data.append(json.loads(baris))

print("JUMLAH DATA GAMBAR:", len(data))

print("\nFITUR DATA:")
print(list(data[0].keys()))

print("\nJUMLAH ANOTASI:")
total_anotasi = sum(len(item["annotation"]) for item in data)
print(total_anotasi)

print("\nLABEL:")
semua_label = []

for item in data:
    for anotasi in item["annotation"]:
        semua_label.extend(anotasi["label"])

print(pd.Series(semua_label).value_counts())

print("\nCONTOH DATA:")
display(pd.DataFrame(data).head())
```

# identifikasi dataset animal classification

```python
import os

folder = "AnimalClassification"

print("ISI FOLDER ANIMALCLASSIFICATION:")
print(os.listdir(folder))
```

```python
from collections import Counter

folder = "AnimalClassification"

ekstensi_gambar = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

jumlah_gambar = 0
jumlah_per_kelas = Counter()

for nama_kelas in os.listdir(folder):
    lokasi_kelas = os.path.join(folder, nama_kelas)

    if os.path.isdir(lokasi_kelas):
        gambar = [
            file for file in os.listdir(lokasi_kelas)
            if file.lower().endswith(ekstensi_gambar)
        ]

        jumlah_per_kelas[nama_kelas] = len(gambar)
        jumlah_gambar += len(gambar)

print("JUMLAH SELURUH GAMBAR:", jumlah_gambar)
print("JUMLAH KELAS HEWAN:", len(jumlah_per_kelas))

print("\nJUMLAH GAMBAR SETIAP KELAS:")
for kelas, jumlah in jumlah_per_kelas.items():
    print(kelas, ":", jumlah)
```

```python
import os
from collections import Counter

folder = "AnimalClassification"

jumlah_gambar = 0
jumlah_per_kelas = Counter()

for nama_hewan in os.listdir(os.path.join(folder, "mammals")):
    lokasi = os.path.join(folder, "mammals", nama_hewan)

    if os.path.isdir(lokasi):
        gambar = [
            file for file in os.listdir(lokasi)
            if file.lower().endswith((".jpg", ".jpeg", ".png", ".webp"))
        ]

        jumlah_per_kelas[nama_hewan] = len(gambar)
        jumlah_gambar += len(gambar)

print("JUMLAH SELURUH GAMBAR:", jumlah_gambar)
print("JUMLAH KATEGORI HEWAN:", len(jumlah_per_kelas))

print("\nJUMLAH GAMBAR SETIAP KATEGORI:")
for hewan, jumlah in jumlah_per_kelas.items():
    print(hewan, ":", jumlah)
```
